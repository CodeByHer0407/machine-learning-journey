### Nominal/OHE Encoding

One hot encoding, also known as nominal encoding, is a technique used to represent categorical data as numerical data, which is more suitable for machine learning algorithms. In this technique, each category is represented as a binary vector where each bit corresponds to a unique category. For example, if we have a categorical variable "color" with three possible values (red, green, blue), we can represent it as one hot encoding as follows:

1. Red: [1, 0, 0]
2. Green: [0, 1, 0]
3. Blue: [0, 0, 1]

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
## Create a dataframe

df = pd.DataFrame(
    {
        "color": ['red', 'blue', 'green', 'green', 'red', 'blue']
    }
)

```

```python
df.head()
```

```python
## create an instance of onehotencoder

encoder = OneHotEncoder()
```

```python
## Perform fit and transform

encoded_values = encoder.fit_transform(df[['color']]).toarray()
```

```python
encoder_df = pd.DataFrame(encoded_values, columns=encoder.get_feature_names_out())
encoder_df
```

```python
## for new data
encoder.transform([['blue']]).toarray()
```

```python
pd.concat([df, encoder_df], axis=1)
```

```python
data = pd.DataFrame({
    'Customer': ['A', 'B', 'C', 'D', 'E'],
    'City': ['Bangalore', 'Delhi', 'Mumbai', 'Bangalore', 'Delhi']
})
```

```python
data.head()
```

```python
data['City'].unique()
```

```python
encoders = OneHotEncoder()
```

```python
encoded_data = encoders.fit_transform(data[['City']]).toarray()
encoded_data
```

```python
encoders.categories_
```

```python
encoders.get_feature_names_out()
```

```python
encoded_df = pd.DataFrame(
    encoded_data,
    columns=encoders.get_feature_names_out(),
)

encoded_df
```

```python
final_df = pd.concat([
    data.drop('City', axis=1), encoded_df],
    axis=1
    )

final_df
```
