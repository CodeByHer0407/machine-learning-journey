## Handling Imbalanced Dataset

An imbalanced dataset occurs when one class has significantly more observations than another class.

### Techniques to handle imbalanced dataset

1. Random oversampling

Oversampling means increasing the number of minority-class samples.



```python
import numpy as np 
import pandas as pd
from sklearn.utils import resample
```

```python
# Set the random seed for reproducibility

np.random.seed(123)


## create a dataframe with two classes
n_samples = 1000
class_0_ratio = 0.9
n_class_0 = int(n_samples * class_0_ratio)
n_class_1 = n_samples - n_class_0
```

```python
n_class_0, n_class_1
```

```python
## create my dataframe with imbalanced dataset

class_0 = pd.DataFrame({
    'feature_1': np.random.normal(loc=0, scale=1, size=n_class_0),
    'feature_2': np.random.normal(loc=0, scale=1, size=n_class_0),
    'target': [0] * n_class_0
})

class_1 = pd.DataFrame({
    'feature_1': np.random.normal(loc=2, scale=1, size=n_class_1),
    'feature_2': np.random.normal(loc=2, scale=1, size=n_class_1),
    'target': [1] * n_class_1
})
```

```python
df = pd.concat([class_0, class_1]).reset_index(drop=True)
```

```python
df['target'].value_counts()
```

```python
## Upsampling

df_minority = df[df['target'] == 1]
df_majority = df[df['target'] == 0]
```

```python
df_minority_unsampled = resample(df_minority, replace=True, n_samples=len(df_majority), random_state=42)
```

```python
df_minority_unsampled.shape
```

```python
df_minority_unsampled.head(5)
```

```python
df_upsampled = pd.concat([df_majority, df_minority_unsampled])
```

```python
df_upsampled['target'].value_counts()
```

```python
df_upsampled = df_upsampled.sample(frac=1, random_state=42).reset_index(drop=True)
```

#### Advantages:

- Simple to implement.
- Doesn't remove existing majority-class data.

#### Disadvantages:

- Duplicates existing records.
- Can cause overfitting because the model repeatedly sees the same minority examples.

2. Random Undersampling

Undersampling means reducing the number of majority-class samples.

```python
## Down Sampling

df['target'].value_counts()
```

```python
df_majority_downsampled = resample(df_majority, replace=False, n_samples=len(df_minority), random_state=42)
```

```python
df_majority_downsampled.shape
```

```python
df_downsampled = pd.concat([df_majority_downsampled, df_minority])
```

```python
df_downsampled.shape
```

```python
df_downsampled['target'].value_counts()
```

```python
df_downsampled.head()
```

```python
df_downsampled = df_downsampled.sample(frac=1,random_state=42).reset_index(drop=True)
```

#### Advantages:

- Smaller dataset.
- Faster model training.
- No duplication of minority records.

#### Disadvantages:

- We lose potentially useful information.
- The model might not learn the full variety of normal transactions.
