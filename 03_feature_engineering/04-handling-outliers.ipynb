## Handling Outliers

An outlier is a data point that is significantly different from the other observations in a dataset.

### 5-Number Summary and Box Plot

```python
## Minimum, Q1, Median, Q3, Maximum, IQR
```

```python
import numpy as np
import pandas as pd
```

```python
lst_marks = [45, 32, 56, 75, 89, 54, 32, 89, 90, 87, 67, 54, 45, 98, 99, 67, 74]
minimum, Q1, median, Q3, maximum = np.quantile(lst_marks, [0, 0.25, 0.50, 0.75, 1.0])
```

```python
print(minimum, Q1, median, Q3, maximum)
```

```python
IQR = Q3 - Q1
print(IQR)
```

```python
lower_fence = Q1 - 1.5*(IQR)
higher_fence =  Q3 + 1.5*(IQR)
```

```python
print(lower_fence, higher_fence)
```

```python
import seaborn as sns
```

```python
sns.boxplot(lst_marks, color='purple')
```

```python
data = pd.DataFrame({
    'salary': [25, 28, 30, 32, 35, 38, 40, 42, 45, 250]
})
```

```python
data
```

```python
minimum, Q1, median, Q3, maximum = np.quantile(data['salary'], [0, 0.25, 0.50, 0.75, 1.0])
print(minimum, Q1, median, Q3, maximum)
```

```python
IQR = Q3 - Q1 
print(IQR)
```

```python
lower_fence = Q1 - 1.5*(IQR)
higher_fence =  Q3 + 1.5*(IQR)
```

```python
print(lower_fence, higher_fence)
```

```python
outliers = data[
    (data['salary'] > higher_fence) | 
    (data['salary'] < lower_fence)]
outliers
```
