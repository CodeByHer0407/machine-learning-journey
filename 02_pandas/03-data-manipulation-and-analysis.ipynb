## Data Manipulation and Analysis with Pandas

Data manipulation and analysis are key tasks in any data science or data analysis project. Pandas provide a wide range of functions for data manipulation and analysis, making it easier to clean, transform and extract data insights from data.

```python
import pandas as pd 
import numpy as np
```

```python
my_df = pd.read_csv('netflix_titles.csv')
```

```python
## Handling missing values

my_df.isnull().any()
```

```python
## Fetch the number of missing values in each column

my_df.isnull().sum()
```

```python
## Fill all the missing values with 0

df_filled = my_df.fillna(0)
```

```python
## filling missing values with the mean of the column

my_df['rating_fillNa'] = my_df['rating'].fillna(my_df['rating'].mode()[0])
```

Create a new column called `rating_fillNa`, which is the same as `rating`, except every missing value is replaced with whatever rating value appears most often.

```python
my_df
```

```python
my_df.dtypes
```

```python
## Renaming columns

my_df.rename(columns={"date_added": "added_date"})
```

```python
## change datatypes

my_df['release_year'] = my_df['release_year'].astype('category') 
```

```python
my_df.dtypes
```

```python
my_df.head(3)
```

```python
## Apply lambda function on columns

my_df['Updated_release_year'] = my_df["release_year"].apply(lambda x: x+5)
```

```python
my_df
```

```python
my_df.dtypes
```

```python
## Data Aggregating and Grouping

grouped_mean = my_df.groupby('type')['release_year'].mean()
print(grouped_mean)
```

Group `my_df` by the `type` column, then take the `release_year`' column, and calculate the mean of it — for each group separately. Store the result in a variable called `grouped_mean`.

```python
my_df.head(5)
```

```python
grouped_sum = my_df.groupby(['country', 'type'])['release_year'].sum()
print(grouped_sum)
```

```python
my_df['country'].nunique()
```

```python
## aggregate multiple function

grouped_agg = my_df.groupby("type")["release_year"].agg(['mean', 'sum', 'count'])
print(grouped_agg)
```

```python
## Merging and joining DataFrames

# create sample DataFrames
df1 = pd.DataFrame({'Key': ['A', 'B', 'C'], 'Value1': [1, 2, 3]})
df2 = pd.DataFrame({'Key': ['A', 'B', 'D'], 'Value2': [4, 5, 6]})
```

```python
## Merge DataFrame on the 'Key Columns'

pd.merge(df1, df2, on="Key", how="inner")
```
