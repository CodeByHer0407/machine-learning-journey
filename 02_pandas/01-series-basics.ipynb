## Pandas-DataFrame And Series

Pandas is a powerful data manipulation library in Python, widely used for data analysis and data cleaning. It provides two primary dat structure: Series and DataFrame. A Series is a one-dimensional array-like object, while a DataFrame is a two-dimensional, size-mutable, and potentially heterogeneous tabular data structure with labeled axes (rows and columns).

```python
import pandas as pd
import numpy as np
```

```python
## Series

## Series: a one-dimensional labeled array holding data of any type such as integers, strings, Python objects etc.(similar to a column in a table)

data = [11, 22, 33, 44, 55]
series = pd.Series(data)
print(f"Series \n", series)
print(type(series))

```

```python
## Creating a series from dictionary

data = {'a': 1, 'b': 2, 'c': 3}
series_dict = pd.Series(data)
print(series_dict)
```

```python
## Constructing series from a dictionary with an index specified

d = {"a": 10, "b": 20, "c": 30}
ser = pd.Series(data=d, index=["a", "b", "c"])
print(ser)

## when the keys of the dictionary don't match the index values, it prints NaN value
ser = pd.Series(data=d, index=["x", "y", "z"])
print(ser)
```

```python
## Constructing series from a list with copy=False

r =[11, 22]
ser = pd.Series(r, copy=False)
print(ser)
```

```python
ser.iloc[0] = 999
print(ser)
print(r)
```

```python
## Constructing series from a 1d arrya with copy=False

r = np.array([1, 2])
indices = ['first', 'second']
ser = pd.Series(r, index=indices, copy=False)
print(r)
print(ser)
```

```python
ser.iloc[0] = 898
print(ser)
print(r)
```

```python
## To create a series with a custom index and view the index labels

cities = ['Kolkata', 'Chicago', 'Toronto', 'Lisbon']
populations = [14.85, 2.71, 2.93, 0.51]
city_series = pd.Series(populations, index=cities)
print(city_series)
print(city_series.index)
```

```python
## To change the index labels of an existing Series

city_series.index = ['KOL', 'CHI', 'TOR', 'LIS']
print(city_series.index)
print(city_series)
```

```python
## return series as arrays depending on the dtype

pd.Series([11, 22, 33]).values
```

```python
pd.Series(list("aabbc")).values
```

```python
## return the data type of the data

s = pd.Series([1, 2, 3])
s.dtype
```

```python
## print summary of the series and returns None

int_values = [1, 2, 3, 4, 5]
text_values = ["alpha", "beta", "gamma", "delta", "epsilon"]
s = pd.Series(text_values, index=int_values)
s.info()
```

```python
## print summary excluding information about its values

s.info(verbose=False)
```

```python
## find the shape of series

s = pd.Series([11, 22, 33])
s.shape
```

```python
## number of dimensions

s.ndim
```

```python
## Transpose of a matrix

s = pd.Series(["Ant", "Bear", "Cow"])
print(s)
```

```python
s.T
```

```python
## check for NaNs

s = pd.Series([1, 2, 3, None])
s.hasnans
```

```python
## check whether index is empty or not

idx = pd.Index([1, 2, 3])
idx
```
