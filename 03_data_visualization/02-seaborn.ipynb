## Data Visualization with Seaborn

Seaborn is a Python visualization library based on Matplotlib that provides a high-level interface for drawing attractive and informative statistical graphics. Seaborn helps in creating complex visualizations with just a few lines of code.

```python
import seaborn as sns
```

```python
## Basic plotting with Seaborn

tips = sns.load_dataset('tips')
tips
```

```python
## create a scatter plot

import matplotlib.pyplot as plt 

sns.scatterplot(x='total_bill', y='tip', data=tips, color="red")
plt.title("Scatter Plot of Total Bill vs Tip")
```

```python
## Line Plot

sns.lineplot(x='size', y='total_bill', data=tips, color='salmon')
plt.title("Line Plot of Total bill by size")
plt.show()
```

```python
## Categorical Plots

sns.barplot(x='day', y='total_bill', data=tips, color="limegreen")
plt.title('Bar Plot of Total Bill by Day')
plt.show()
```

```python
## Box plot

sns.boxplot(x='day', y='total_bill', data=tips, color='coral')
```

```python
## Violin Plot

sns.violinplot(x='day', y='total_bill', data=tips, color='skyblue')
```

```python
## Histograms

sns.histplot(tips['total_bill'], bins=10, kde=True, color='steelblue')
```

```python
## KDE Plot

sns.kdeplot(tips['total_bill'], fill=True, color='pink')
```

```python
## Pairplot

sns.pairplot(tips)
```

```python
## Heatmap

corr = tips[['total_bill', 'tip', 'size']].corr()
corr
```

```python
sns.heatmap(corr, annot=True, cmap='coolwarm')
```
