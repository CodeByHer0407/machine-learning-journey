2. Create a NumPy array of shape (5, 5) with random integers. Extract the elements on the border.

```python
val1 = np.random.default_rng()
my_val = val1.random(25).reshape(5,5)
print(my_val)

top = my_val[0, :]
bottom = my_val[-1, :]
left = my_val[1:-1,0]
right = my_val[1:-1, -1]

print(f"TOP: ", top)
print(f"BOTTOM: ",bottom)
print(f"LEFT: ", left)
print(f"RIGHT: ", right)

border_elements = np.concatenate([top, bottom, left, right])
print(f"Border elements: ", border_elements)
```

### Assignment 4: Statistical Operations

1. Create a NumPy array of shape (5, 5) filled with random integers. Compute the mean, median, standard deviation, and variance of the array.

```python
my_arr3 = np.random.randint(1, 21, size=(5, 5))
print(my_arr3)


mean = np.mean(my_arr3)
median = np.median(my_arr3)
std_dev = np.std(my_arr3)
variance = np.var(my_arr3)

print(f"Mean: {mean}")
print(f"Median: {median}")
print(f"Standard Deviation: {std_dev}")
print(f"Variance: {variance}")
```

2. Create a NumPy array of shape (3, 3) with values from 1 to 9. Normalize the array (i.e., scale the values to have a mean of 0 and a standard deviation of 1).

```python
my_arr4 = np.arange(1, 10).reshape(3, 3)
print(my_arr4)

mean = np.mean(my_arr4)
std_dev = np.std(my_arr4)
normalized_array = (my_arr4 - mean) / std_dev
print(f'Normalized array: \n', normalized_array)
```

### Assignment 6: Linear Algebra

1. Create a NumPy array of shape (3, 3) representing a matrix. Compute its determinant, inverse, and eigenvalues.

```python
matrix = np.random.randint(1, 11, size=(3,3))
print("Original matrix: \n", matrix)

determinant = np.linalg.det(matrix)
print(f"Determinant: ", determinant)

transposed_matrix = np.transpose(matrix)
print(f"Transposed matrix: ", transposed_matrix)

inverse = np.linalg.inv(matrix)
print(f"Inverse of a matrix: ", inverse)

eigens = np.linalg.eigvals(matrix)
print(f"Eigen values: ", eigens)
```

### Assignment 9: Structured Arrays

1. Create a structured array with fields 'name' (string), 'age' (integer), and 'weight' (float). Add some data and sort the array by age.

```python
dtype = [('name', 'U10'), ('age', 'i8'), ('weight', 'f8')]

data = [('Alice', 30, 58.6), ('Bob', 25, 68.9), ('Charlie', 32, 63)]

structured_array = np.array(data, dtype)
print(f"Structured Array:\n", structured_array)

sorted_data = np.sort(structured_array, order='age')
print(f"Sorted array by age: ", sorted_data)
```

2. Create a structured array with fields 'x' and 'y' (both integers). Add some data and compute the Euclidean distance between each pair of points.

```python
dtype = [('x', 'i8'), ('y', 'i8')]
data = np.array([(1, 2), (3, 4), (5, 6)], dtype=dtype)

distances = np.sqrt((data['x'][:, np.newaxis] - data['x'])**2 + (data['y'][:, np.newaxis] - data['y'])**2)
print("Euclidean distances:")
print(distances)
```

### Assignment 10: Masked Arrays

1. Create a masked array of shape (4, 4) with random integers and mask the elements greater than 10. Compute the sum of the unmasked elements.

```python
import numpy.ma as ma 

array = np.random.randint(1, 21, size=(4, 4))
masked_array = ma.masked_greater(array, 10)
print("Original array:")
print(array)
print("Masked array:")
print(masked_array)

sum_unmasked = masked_array.sum()
print("Sum of unmasked elements:", sum_unmasked)
```

2. Create a masked array of shape (3, 3) with random integers and mask the diagonal elements. Replace the masked elements with the mean of the unmasked elements.

```python
array = np.random.randint(1, 21, size=(3, 3))
masked_array = ma.masked_array(array, mask=np.eye(3, dtype=bool))
print("Original array:")
print(array)
print("Masked array:")
print(masked_array)


mean_unmasked = masked_array.mean()
masked_array = masked_array.filled(mean_unmasked)
print("Modified masked array:")
print(masked_array)
```
